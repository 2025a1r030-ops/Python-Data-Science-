**Task:** Extract GameStop (`GME`) historical stock data using the `yfinance` library, reset the index, and display the first five rows using `head()`.

### Step 1: Import Required Libraries
Import the `yfinance` library to fetch historical market data.

```python
# Install yfinance if not already installed in your environment
# !pip install yfinance

# Import yfinance library
import yfinance as yf
```

### Step 2: Create GameStop Ticker Object & Extract Historical Stock Data
Using the `Ticker` function, create a ticker object for GameStop (`GME`) and extract the maximum available historical data using `history(period="max")`.

```python
# Create a ticker object for GameStop (GME)
gme = yf.Ticker("GME")

# Extract historical stock data for the maximum available period
gme_data = gme.history(period="max")
```

### Step 3: Reset the Index and Display the First Five Rows
Reset the index of the `gme_data` DataFrame using `inplace=True` and display the first five rows using `head()`.

```python
# Reset the index of the DataFrame
gme_data.reset_index(inplace=True)

# Display the first five rows of gme_data
gme_data.head()
```
