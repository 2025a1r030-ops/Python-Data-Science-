# Question 4: Use Webscraping to Extract GME Revenue Data

**Course:** Coursera IBM Python Project for Data Science – Final Project  
**Task:** Extract GameStop (`GME`) quarterly revenue data using web scraping (`requests` and `BeautifulSoup`), clean the data, and display the last five rows of the `gme_revenue` DataFrame using `tail()`.

### Step 1: Import Required Libraries
Import `requests`, `BeautifulSoup` from `bs4`, and `pandas` to fetch and parse the webpage HTML.

```python
# Import required libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd
import warnings

# Ignore future warnings for clean output
warnings.filterwarnings("ignore", category=FutureWarning)
```

### Step 2: Download the Webpage Using `requests`
Use the `requests.get()` function to download the official IBM Skills Network webpage containing GameStop historical revenue data.

```python
# URL containing GameStop revenue data provided by IBM Coursera lab
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/stock.html"

# Send GET request and save response text
html_data = requests.get(url).text
```

### Step 3: Parse the HTML Data Using `BeautifulSoup`

```python
# Parse HTML data using BeautifulSoup
soup = BeautifulSoup(html_data, "html.parser")
```

### Step 4: Extract GameStop Quarterly Revenue Table into `gme_revenue` DataFrame
Locate the **GameStop Quarterly Revenue** table in the parsed HTML, extract the `Date` and `Revenue` columns, and clean the `Revenue` column by removing dollar signs (`$`), commas (`,`), and any empty/null rows.

```python
# Initialize an empty DataFrame with Date and Revenue columns
gme_revenue = pd.DataFrame(columns=["Date", "Revenue"])

# Find the table containing GameStop Quarterly Revenue
for table in soup.find_all("table"):
    if "GameStop Quarterly Revenue" in table.text:
        for row in table.find("tbody").find_all("tr"):
            col = row.find_all("td")
            if col:
                date = col[0].text.strip()
                revenue = col[1].text.strip()
                gme_revenue = pd.concat(
                    [gme_revenue, pd.DataFrame({"Date": [date], "Revenue": [revenue]})],
                    ignore_index=True
                )

# Remove comma and dollar sign from the Revenue column
gme_revenue["Revenue"] = gme_revenue["Revenue"].str.replace(r",|\$", "", regex=True)

# Drop null or empty string rows
gme_revenue.dropna(inplace=True)
gme_revenue = gme_revenue[gme_revenue["Revenue"] != ""]
```

### Step 5: Display the Last Five Rows Using `tail()`

```python
# Display the last five rows of the gme_revenue DataFrame
gme_revenue.tail()
```
