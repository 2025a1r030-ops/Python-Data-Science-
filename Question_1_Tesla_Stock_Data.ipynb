# Question 1: Use yfinance to Extract Stock Data

Extract Tesla stock data using the `yfinance` library. Reset the index of `tesla_data` and display its first five rows using `head()`.

```python
# Import the yfinance library
import yfinance as yf
```

### Create a ticker object for Tesla

```python
# Create the Tesla ticker object
tesla = yf.Ticker("TSLA")
```

### Extract Tesla historical stock data

```python
# Extract historical stock data for the maximum available period
tesla_data = tesla.history(period="max")
```

### Reset the index and display the first five rows

```python
# Reset the index and save the updated dataframe
tesla_data.reset_index(inplace=True)

# Display the first five rows
tesla_data.head()
```
