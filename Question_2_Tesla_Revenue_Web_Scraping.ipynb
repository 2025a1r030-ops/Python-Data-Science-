# Question 2: Use Webscraping to Extract Tesla Revenue Data

Use Python web-scraping libraries to collect Tesla revenue data, clean the Revenue column, and display the last five rows of `tesla_revenue` using `tail()`.

```python
# Import the required libraries
import pandas as pd
import requests
from bs4 import BeautifulSoup
from io import StringIO
```

### Retrieve the Tesla revenue webpage

```python
# URL containing Tesla revenue history
url = "https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue"

# Request the webpage
headers = {"User-Agent": "Mozilla/5.0"}
html_data = requests.get(url, headers=headers, timeout=30)
html_data.raise_for_status()
soup = BeautifulSoup(html_data.text, "html.parser")
```

### Extract and clean the Tesla revenue table

The code below locates the table containing Date and Revenue columns, then cleans the revenue values.

```python
# Read the tables present in the webpage
tables = pd.read_html(StringIO(str(soup)))

# Find the table that contains Date and Revenue columns
tesla_revenue = None
for table in tables:
    table.columns = [str(col).strip() for col in table.columns]
    if 'Date' in table.columns and 'Revenue' in table.columns:
        tesla_revenue = table[['Date', 'Revenue']].copy()
        break

if tesla_revenue is None:
    raise ValueError('Could not find a Tesla revenue table with Date and Revenue columns.')

# Remove dollar signs and commas from revenue values
tesla_revenue['Revenue'] = tesla_revenue['Revenue'].astype(str).str.replace(r'[\$,]', '', regex=True)

# Remove empty or missing revenue rows
tesla_revenue.dropna(inplace=True)
tesla_revenue = tesla_revenue[tesla_revenue['Revenue'] != '']
tesla_revenue.reset_index(drop=True, inplace=True)
```

### Display the last five rows of Tesla revenue data

```python
# Display the last five rows
tesla_revenue.tail()
```
