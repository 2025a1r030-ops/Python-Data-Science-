# Final Assignment: Part 1 - Create Visualizations using Matplotlib, Seaborn & FoliumThis notebook covers all tasks required to analyze automobile sales data and visualize trends during recession and non-recession periods.

```python
import numpy as npimport pandas as pd%matplotlib inlineimport matplotlib.pyplot as pltimport seaborn as snsimport foliumprint('Libraries imported successfully!')
```

```python
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data/historical_automobile_sales.csv"df = pd.read_csv(URL)print('Data downloaded and read into a dataframe!')print(df.head())
```

### Task 1.1: Line chart to show how automobile sales fluctuate from year to year.

```python
df_line = df.groupby('Year')['Automobile_Sales'].mean().reset_index()plt.figure(figsize=(10, 6))plt.plot(df_line['Year'], df_line['Automobile_Sales'], marker='o', linestyle='-', color='b')plt.xlabel('Year')plt.ylabel('Average Automobile Sales')plt.title('Automobile Sales Fluctuation from Year to Year')plt.grid(True)plt.show()
```

### Task 1.2: Line chart to show trends in advertising expenditure correlate with automobile sales during non-recession periods.

```python
non_recession_df = df[df['Recession'] == 0]plt.figure(figsize=(10, 6))plt.plot(non_recession_df['Year'], non_recession_df['Advertising_Expenditure'], label='Advertising Expenditure', marker='s', color='r')plt.plot(non_recession_df['Year'], non_recession_df['Automobile_Sales'], label='Automobile Sales', marker='o', color='b')plt.xlabel('Year')plt.ylabel('Amount / Sales')plt.title('Advertising Expenditure vs Automobile Sales during Non-Recession Periods')plt.legend()plt.grid(True)plt.show()
```

### Task 1.3: Bar chart to compare the sales trend per vehicle type for a recession period with a non-recession period.

```python
plt.figure(figsize=(12, 6))sns.barplot(x='Vehicle_Type', y='Automobile_Sales', hue='Recession', data=df, ci=None)plt.xlabel('Vehicle Type')plt.ylabel('Average Automobile Sales')plt.title('Comparison of Sales Trend per Vehicle Type (Recession vs Non-Recession)')plt.legend(title='Recession', labels=['Non-Recession', 'Recession'])plt.xticks(rotation=45)plt.show()
```

### Task 1.4: Subplots to compare the variations in GDP during recession and non-recession periods by developing line plots for each period.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))rec_data = df[df['Recession'] == 1]non_rec_data = df[df['Recession'] == 0]sns.lineplot(x='Year', y='GDP', data=rec_data, ax=axes[0], color='red', marker='o')axes[0].set_title('GDP Variation during Recession Period')axes[0].set_xlabel('Year')axes[0].set_ylabel('GDP')sns.lineplot(x='Year', y='GDP', data=non_rec_data, ax=axes[1], color='blue', marker='o')axes[1].set_title('GDP Variation during Non-Recession Period')axes[1].set_xlabel('Year')axes[1].set_ylabel('GDP')plt.tight_layout()plt.show()
```

### Task 1.5: Bubble plot for displaying the impact of seasonality on Automobile Sales.

```python
plt.figure(figsize=(10, 6))sns.scatterplot(data=df, x='Month', y='Automobile_Sales', size='Advertising_Expenditure', hue='Seasonality_Weight', palette='viridis', sizes=(20, 200), alpha=0.7)plt.xlabel('Month')plt.ylabel('Automobile Sales')plt.title('Seasonality Impact on Automobile Sales')plt.legend(bbox_to_anchor=(1.05, 1), loc=2)plt.show()
```

### Task 1.6: Scatter plot to identify the correlation between average vehicle price relate to the sales volume during recessions.

```python
plt.figure(figsize=(10, 6))sns.scatterplot(data=rec_data, x='Price', y='Automobile_Sales', color='purple')plt.xlabel('Average Vehicle Price')plt.ylabel('Automobile Sales Volume')plt.title('Correlation between Average Vehicle Price and Sales Volume during Recessions')plt.grid(True)plt.show()
```

### Task 1.7: Pie chart to display the portion of advertising expenditure of XYZAutomotives during recession and non-recession periods.

```python
exp_rec = df.groupby('Recession')['Advertising_Expenditure'].sum()plt.figure(figsize=(8, 8))plt.pie(exp_rec, labels=['Non-Recession', 'Recession'], autopct='%1.1f%%', startangle=90, colors=['skyblue', 'salmon'])plt.title('Portion of Advertising Expenditure: Recession vs Non-Recession')plt.show()
```

### Task 1.8: Pie chart to display the total Advertisement expenditure for each vehicle type during the recession period.

```python
exp_vehicle = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()plt.figure(figsize=(8, 8))plt.pie(exp_vehicle, labels=exp_vehicle.index, autopct='%1.1f%%', startangle=90)plt.title('Total Advertisement Expenditure per Vehicle Type during Recession')plt.show()
```

### Task 1.9: Line chart to analyze the effect of the unemployment rate on vehicle type and sales during the Recession Period.

```python
plt.figure(figsize=(12, 6))sns.lineplot(data=rec_data, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type', marker='o')plt.xlabel('Unemployment Rate')plt.ylabel('Automobile Sales')plt.title('Effect of Unemployment Rate on Vehicle Type and Sales during Recession Period')plt.legend(title='Vehicle Type')plt.grid(True)plt.show()
```
